## **Exercício 1**

Defina uma classe chamada Flor, com três atributos privados: *nome* (nome da flor, string), *petalas* (número de pétalas, int) e *preco* (preço da flor, float).

1) A classe deverá incluir um método construtor para inicializar as variáveis e métodos get e set básicos (sem decoradores) para, respectivamente, recuperar e atribuir valores para estes atributos.

2) Tanto o construtor como os *setters* devem validar os tipos de cada dado de entrada e imprimir uma mensagem de erro caso o tipo errado seja especificado. Além disso, se o tipo errado for especificado, o valor *None* deverá ser atribuído ao atributo.

3) Depois de definir a classe, instancie um objeto “f”, da classe Flor, com as seguintes características: “margarida”, 13 pétalas, R$ 85,90. Imprima o valor dos três atributos usando os getters.

4) Por fim, utilize os *setters* adequados para alterar o número de pétalas para 21 e o preço para R$ 99,90. Imprima o valor dos três atributos usando os getters.

```python
# classe Flor
class Flor:

  # Método Construtor
  def __init__(self, nome, petalas, preco):
    # self._nome = nome
    self.set_nome(nome)
    # self._petalas = petalas
    self.set_petalas(petalas)
    # self._preco = preco
    self.set_preco(preco)

  # Atributo 1: nome
  # Getter do atributo nome sem o decorador (@property)
  def get_nome(self):
    return self._nome

  # Setter do atributo nome sem o decorador (@nome.self)
  def set_nome(self, nome):
    if (type(nome) == str):
      self._nome = nome
    else:
      self._nome = None
      print('Nome inválido! Deve ser uma string.')


  # Atributo 2: petalas
  # Getter do atributo petalas sem o decorador (@property)
  def get_petalas(self):
    return self._petalas

  # Setter do atributo petalas sem o decorador (@petalas.self)
  def set_petalas(self, petalas):
    if (type(petalas) == int):
      self._petalas = petalas
    else:
      self._petalas = None
      print('Quantidade inválida! Deve ser um número inteiro.')


  # Atributo 3: preco
  # Getter do atributo preco sem o decorador (@property)
  def get_preco(self):
    return self._preco

  # Setter do atributo preco sem o decorador (@preco.self)
  def set_preco(self, preco):
    if (type(preco) == float):
      self._preco = preco
    else:
      self._preco = None
      print('Valor inválido! Deve ser um número real.')
```

A classe Flor foi criada com a declaração do método construtor usando setters da própria classe para verificação pedida no exercício.

Caso o nome passado não esteja no formato de string (str) então uma mensagem informará que o formato inserido é inválido, e também atribuirá o valor None ao atributo.

Os três atributos de instância se enquadram como atributos protegidos porque são instanciados usando um underscore. É uma prática para indicar que não devem ser alterados ou que é preciso ter cautela no acesso a eles.

```python
# Instanciar em f uma margarida com 13 pétalas que custa R$ 89,90.
f = Flor("margarida", 13, 89.90)

print("Nome: ", f.get_nome())
print("Pétalas: ", f.get_petalas())
print("Preço: R$", f.get_preco())
```

```python
# Alterar as propriedades de f para obtermos uma margarida com mais pétalas (21) e mais cara (R$ 99,90)
f.set_petalas(21)
f.set_preco(99.90)

print("Nome: ", f.get_nome())
print("Pétalas: ", f.get_petalas())
print("Preço: R$", f.get_preco())
```

```python
rosa = Flor(nome = 'Rosa', petalas = 20, preco = 10.99)

# A classe tem atributos protegidos (com um underscore)
# Para acessar os atributos é preciso usar underscore
print(rosa._nome)
# Ou podemos usar o método criado para retornar o atributo
print(rosa.get_nome())

print(rosa.get_petalas())
print(rosa.get_preco())
```

## **Exercício 2**

Crie uma classe chamada *FlorNova* com a mesma definição da classe *Flor*, do exercício anterior, mas agora utilizando **decoradores** nos *getters* e *setters*.

```python
class FlorNova:

  def __init__(self, nome, petalas, preco):
    self.nome = nome
    self.petalas = petalas
    self.preco = preco

  # Atributo 1: nome
  # Getter com decorador (@property)
  @property
  def nome(self):
    return self._nome

  # Setter com decorador (@nome.setter)
  @nome.setter
  def nome(self, nome):
    if type(nome) == str:
       self._nome = nome
    else:
      self._nome = None
      print('Nome inválido! Deve ser do tipo string')


  # Atributo 2: petalas
  # Getter com decorador (@property)
  @property
  def petalas(self):
    return self._petalas

  # Setter com decorador (@petalas.setter)
  @petalas.setter
  def petalas(self, petalas):
    if type(petalas) == int:
      self._petalas = petalas
    else:
      self._petalas = None
      print('Quantidade inválida! Deve ser um número inteiro.')


  # Atributo 3: preco
  # Getter com decorador (@property)
  @property
  def preco(self):
    return self._preco

  # Setter com decorador (@preco.setter)
  @preco.setter
  def preco(self, preco):
    if type(preco) == float:
      self._preco = preco
    else:
      self._preco = None
      print('Valor inválido! Deve ser um número real.')

```

Importante notar que os atributos de instância (nome, petalas e preco) continuam sendo protegidos (criados com um underscore), mas isso não é mais evidente ao olhar o método construtor e sim o Setter de cada atributo.

```python
rosa = FlorNova('Rosa', 20, 100.0)
```

```python
print('Nome: ', rosa.nome)
print('Pétalas: ', rosa.petalas)
print('Preço: R$', rosa.preco)
```

```python
rosa.petalas = 10
rosa.preco = 50.0

print('Nome: ', rosa.nome)
print('Pétalas: ', rosa.petalas)
print('Preço: R$', rosa.preco)
```

## **Encapsulamento e a Redução da Congeneridade**

A **congeneridade** é a medida de quanto a alteração em uma parte do sistema obriga a modificação de outra. Quando deixamos os atributos expostos, o código externo fica "amarrado" à estrutura interna da classe.

Ao utilizarmos o encapsulamento com `@property` e `@setter`, quebramos essa dependência:

- **Independência de Implementação**: Podemos alterar o nome do atributo interno (ex: de `_preco` para `_valor_venda`) ou adicionar regras de validação sem que ninguém precise mudar uma única linha do código que usa a classe.
   
- **Manutenção Facilitada**: Reduzimos o "efeito cascata" de erros, pois as regras de negócio ficam centralizadas em um só lugar.

- **Interface Estável**: O usuário da classe sabe _o que_ pode ser feito (acessar o preço), mas não precisa saber _como_ isso é processado internamente.
